```python
import requests
import pandas as pd

# Obtener datos de la API
url = "https://api.frankfurter.dev/v1/latest?from=EUR"
respuesta = requests.get(url)
datos = respuesta.json()

# Crear una fila con todos los cambios
fila = datos["rates"]
fila["date"] = datos["date"]

# Crear DataFrame
df = pd.DataFrame([fila])

# Guardar CSV
df.to_csv("cambio_moneda.csv", index=False)

# Mostrar datos
display(df)
```

```python
import pandas as pd

url_csv = "https://github.com/leiregarridogonzalez0-droid/cambio_moneda.csv.git"

df = pd.read_csv(url_csv)

print("Datos del CSV:")
display(df)

print("Estadísticos:")
display(df.describe())
```

```python
import matplotlib.pyplot as plt

# Monedas que queremos representar
monedas = ["USD", "GBP", "JPY", "CHF"]

# Valores de cada moneda
valores = [df["USD"].iloc[0], df["GBP"].iloc[0], df["JPY"].iloc[0], df["CHF"].iloc[0]]

# Crear el gráfico
plt.bar(monedas, valores)

plt.title("Cambio del EUR frente a otras monedas")
plt.xlabel("Moneda")
plt.ylabel("Cambio")

plt.show()
```
